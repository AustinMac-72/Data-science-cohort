### Step 2: Implement Price Action Strategy & Backtesting
We will download data for a popular asset (e.g., Apple - `AAPL`) and build a Simple Moving Average (SMA) crossover strategy, which is a foundational price action technique.

```python
import yfinance as yf
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Download historical data
ticker = 'AAPL'
data = yf.download(ticker, start='2020-01-01', end='2023-12-31')

# Clean multi-index columns if present
if isinstance(data.columns, pd.MultiIndex):
    data.columns = data.columns.droplevel(1)

# Define SMA windows
short_window = 20
long_window = 50

# Calculate Moving Averages
data['SMA_Short'] = data['Close'].rolling(window=short_window).mean()
data['SMA_Long'] = data['Close'].rolling(window=long_window).mean()

# Generate signals: 1 when Short SMA > Long SMA (Bullish), else 0
data['Signal'] = 0.0
data['Signal'] = np.where(data['SMA_Short'] > data['SMA_Long'], 1.0, 0.0)

# Calculate daily positions (changes in signal)
data['Position'] = data['Signal'].diff()

# Calculate Daily Returns
data['Market_Returns'] = data['Close'].pct_change()

# Calculate Strategy Returns (Shift signal by 1 day to prevent lookahead bias)
data['Strategy_Returns'] = data['Market_Returns'] * data['Signal'].shift(1)

# Calculate Cumulative Returns
data['Cumulative_Market'] = (1 + data['Market_Returns'].fillna(0)).cumprod() - 1
data['Cumulative_Strategy'] = (1 + data['Strategy_Returns'].fillna(0)).cumprod() - 1

display(data[['Close', 'SMA_Short', 'SMA_Long', 'Signal']].tail(10))
```

### Step 3: Visualize Performance
Let's plot the asset price with buy/sell signals alongside the strategy's cumulative returns compared to the buy-and-hold market return.

```python
fig, (ax1, ax2) = plt.subplots(2, 1, figsize=(14, 10), sharex=True)

# Plot Price and Moving Averages
ax1.plot(data.index, data['Close'], label='Close Price', color='gray', alpha=0.6)
ax1.plot(data.index, data['SMA_Short'], label=f'{short_window}-day SMA', color='blue', alpha=0.8)
ax1.plot(data.index, data['SMA_Long'], label=f'{long_window}-day SMA', color='orange', alpha=0.8)

# Plot Buy Signals
ax1.plot(data[data['Position'] == 1].index,
         data['SMA_Short'][data['Position'] == 1],
         '^', markersize=10, color='g', label='Buy Signal', lw=0)

# Plot Sell Signals
ax1.plot(data[data['Position'] == -1].index,
         data['SMA_Short'][data['Position'] == -1],
         'v', markersize=10, color='r', label='Sell Signal', lw=0)

ax1.set_title(f'{ticker} Price Action & Signals')
ax1.set_ylabel('Price ($)')
ax1.legend()
ax1.grid(True, linestyle='--', alpha=0.5)

# Plot Cumulative Returns
ax2.plot(data.index, data['Cumulative_Market'] * 100, label='Market Returns (Buy & Hold)', color='gray')
ax2.plot(data.index, data['Cumulative_Strategy'] * 100, label='Strategy Returns', color='green')
ax2.set_title('Cumulative Performance Comparison')
ax2.set_xlabel('Date')
ax2.set_ylabel('Returns (%)')
ax2.legend()
ax2.grid(True, linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```
